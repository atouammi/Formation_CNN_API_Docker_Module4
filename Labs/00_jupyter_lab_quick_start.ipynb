## 7. Notions de base sur le noyau

Le **noyau** (kernel) est le processus Python qui exécute votre code.

Actions utiles :

- **Interrupt Kernel** : arrêter une cellule qui tourne trop longtemps.

- **Restart Kernel and Run All Cells** : réexécuter le notebook depuis le début.

Si un notebook se comporte bizarrement, la première solution est généralement :

1. Enregistrer le notebook.
2. Redémarrer le noyau.
3. Réexécuter les cellules depuis le haut.

# Configurer Jupyter et l'environnement d'exécution

Cette partie s'adresse aux formateurs, aux assistants et aux participants avancés qui ont besoin de comprendre comment l'environnement de TP est démarré, configuré, vérifié et réparé.


## 12. Comment l'environnement de TP est déployé

Dans ce cours, les participants ouvrent JupyterLab via une URL dans leur navigateur. Derrière la page, il y a généralement trois éléments :

- **Navigateur** : l'interface du participant.
- **Serveur Jupyter** : le service web qui gère les notebooks, les noyaux, les terminaux et les fichiers.
- **Environnement Python / noyau** : le processus qui exécute réellement le code Python.

Une commande de démarrage typique ressemble à ceci :

```bash
jupyter lab \
  --notebook-dir=/path/to/course-notebooks \
  --ip=0.0.0.0 \
  --port=8888 \
  --no-browser \
  --allow-root
```

Options importantes :

- `--notebook-dir` : le dossier que les participants voient après connexion.
- `--ip=0.0.0.0` : autorise l'accès depuis un navigateur distant.
- `--port=8888` : le port du service Jupyter.
- `--no-browser` : n'ouvre pas de navigateur local sur le serveur.
- `--allow-root` : nécessaire uniquement si Jupyter est lancé en tant que root.

Pour une vraie session de formation, gardez la protection par mot de passe activée et n'exposez jamais Jupyter sans contrôle d'accès.


## 13. Mot de passe, jeton et contrôle d'accès

Jupyter peut utiliser un jeton (token) ou un mot de passe.

- **Mode jeton** : Jupyter affiche une longue URL de connexion au démarrage.
- **Mode mot de passe** : les participants se connectent avec le mot de passe du cours.

Pour une formation, le mode mot de passe est plus simple pour les participants, mais le mot de passe ne doit être communiqué qu'à eux.

Commandes de configuration courantes :

```bash
# Générer ou modifier la configuration du serveur Jupyter
jupyter server --generate-config

# Définir un mot de passe de façon interactive
jupyter server password

# Lister les serveurs Jupyter actifs
jupyter server list
```

Rappels pratiques :

- Ne mettez pas de mots de passe dans les notebooks.
- Ne collez pas les identifiants du serveur dans un chat public ou dans des diapositives.
- Changez le mot de passe après la formation si le serveur reste en ligne.


## 14. Notions de base sur l'environnement Python

Un notebook s'exécute dans un environnement Python. C'est l'environnement qui détermine quelles bibliothèques, et quelles versions, sont disponibles.

Outils courants :

- `conda` : gère les environnements Python et les paquets binaires.
- `pip` : installe des paquets Python depuis PyPI.
- `requirements.txt` : liste les dépendances pip.
- `environment.yml` : liste les dépendances d'un environnement conda.

Bonnes pratiques pour des TP :

1. Préparer l'environnement avant la séance.
2. Noter les versions exactes.
3. Tester chaque notebook de haut en bas.
4. Garder une sauvegarde propre ou une image de l'environnement.
5. Éviter de faire installer des bibliothèques GPU lourdes pendant la séance.


````python
# Vérifier l'exécutable Python actif et sa version.
import sys
import platform

print("Exécutable Python :", sys.executable)
print("Version de Python :", sys.version)
print("Plateforme :", platform.platform())
````

````python
# Vérifier les versions des principales bibliothèques utilisées dans ce cours.
packages = [
    "numpy",
    "pandas",
    "scipy",
    "sklearn",
    "matplotlib",
    "plotly",
    "tensorflow",
    "torch",
]

import importlib

for package in packages:
    try:
        module = importlib.import_module(package)
        version = getattr(module, "__version__", "version inconnue")
        print(f"{package}: {version}")
    except Exception as exc:
        print(f"{package}: non disponible ({exc.__class__.__name__})")
````

## 17. Vérifier le GPU

La disponibilité du GPU dépend du serveur, du pilote NVIDIA, des bibliothèques CUDA et des versions des frameworks.

Faites ces vérifications avant d'exécuter des notebooks gourmands en GPU :

- `nvidia-smi` : confirme que le GPU et le pilote sont visibles.
- Test CUDA de PyTorch : confirme que PyTorch peut utiliser le GPU.
- Test GPU de TensorFlow : confirme que TensorFlow voit les GPU.

Un GPU peut être visible pour un framework et pas pour un autre si les versions de CUDA/cuDNN sont incompatibles.


````python
# Vérification du GPU avec PyTorch.
try:
    import torch
    print("Version de Torch :", torch.__version__)
    print("CUDA disponible :", torch.cuda.is_available())
    if torch.cuda.is_available():
        print("Nombre de GPU :", torch.cuda.device_count())
        print("Nom du GPU :", torch.cuda.get_device_name(0))
except Exception as exc:
    print("Échec de la vérification PyTorch :", repr(exc))
````

````python
# Vérification des périphériques avec TensorFlow.
try:
    import tensorflow as tf
    print("Version de TensorFlow :", tf.__version__)
    print("GPU physiques :", tf.config.list_physical_devices("GPU"))
except Exception as exc:
    print("Échec de la vérification TensorFlow :", repr(exc))
````

## 19. Résoudre les problèmes d'environnement

**Problème : bibliothèque introuvable**

- Vérifiez le noyau actif.
- Installez la bibliothèque dans l'environnement du noyau courant avec `%pip install ...`.
- Redémarrez le noyau.

**Problème : les résultats du notebook semblent obsolètes**

- Utilisez **Kernel -> Restart Kernel and Run All Cells**.

**Problème : le GPU n'est pas utilisé**

- Vérifiez `nvidia-smi` sur le serveur.
- Vérifiez les versions des frameworks.
- Confirmez que le notebook utilise le bon environnement.
- Repassez sur CPU si l'objectif de la séance est conceptuel plutôt que lié aux performances.

**Problème : la page ne parvient pas à se connecter au noyau**

- Rafraîchissez le navigateur.
- Vérifiez que le serveur Jupyter tourne toujours.
- Redémarrez Jupyter si nécessaire.

**Problème : des participants modifient les notebooks par erreur**

- Gardez une copie de sauvegarde.
- Fournissez une archive propre.
- Conservez les versions vérifiées par le formateur dans `executed/` ou `backups/`.


````python
# Rapport d'environnement compact pour le dépannage.
import sys
import importlib
from pathlib import Path

report = {
    "python": sys.version.split()[0],
    "executable": sys.executable,
    "cwd": str(Path.cwd()),
}

for package in ["numpy", "pandas", "sklearn", "matplotlib", "torch", "tensorflow"]:
    try:
        module = importlib.import_module(package)
        report[package] = getattr(module, "__version__", "inconnue")
    except Exception:
        report[package] = "non disponible"

report
````
